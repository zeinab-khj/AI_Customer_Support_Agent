```python
from datasets import load_dataset

dataset = load_dataset(
    "bitext/Bitext-customer-support-llm-chatbot-training-dataset"
)
```

```python
dataset
dataset.column_names
dataset["train"].features
```

```python
dataset["train"][:5]
```

```python
import pandas as pd

df = dataset["train"].to_pandas()

print(df.shape)
display(df.head(10))
```

```python
print(df.info())
```

```python
display(df.describe(include="all").T)
```

```python
print("Categories:")
display(df["category"].value_counts())

print("\nIntents:")
display(df["intent"].value_counts())
```

```python
print("Category percentages:")
display(
    df["category"].value_counts(normalize=True).mul(100).round(2)
)

print("\nIntent percentages:")
display(
    df["intent"].value_counts(normalize=True).mul(100).round(2)
)
```

```python
category_counts = df["category"].value_counts()
intent_counts = df["intent"].value_counts()
print("Category imbalance ratio:",
      category_counts.max() / category_counts.min())

print("Intent imbalance ratio:",
      intent_counts.max() / intent_counts.min())
```

```python
duplicates = (
    df[df["instruction"].duplicated(keep=False)]
    .sort_values("instruction")
)

display(duplicates.head(30))
```

```python
print(
    "Number of duplicated instruction rows:",
    df["instruction"].duplicated().sum()
)

print(
    "Number of duplicated instruction values:",
    df["instruction"].value_counts().gt(1).sum()
)
```

```python
instruction_intent_counts = (
    df.groupby("instruction")["intent"]
      .nunique()
)

print(
    "Instructions with multiple intents:",
    (instruction_intent_counts > 1).sum()
)
```

```python
instruction_category_counts = (
    df.groupby("instruction")["category"]
      .nunique()
)

print(
    "Instructions with multiple categories:",
    (instruction_category_counts > 1).sum()
)
```

```python
df["instruction_chars"] = df["instruction"].str.len()
df["response_chars"] = df["response"].str.len()

display(
    df[["instruction_chars", "response_chars"]].describe(
        percentiles=[0.5, 0.75, 0.90, 0.95, 0.99]
    )
)
```

```python
display(
    df.nlargest(10, "response_chars")[
        ["instruction", "response", "response_chars"]
    ]
)
```

```python
duplicate_instructions = (
    df[df["instruction"].duplicated(keep=False)]
    .groupby("instruction")
    .agg(
        n_rows=("instruction", "size"),
        n_responses=("response", "nunique"),
        n_intents=("intent", "nunique"),
        n_categories=("category", "nunique")
    )
    .sort_values("n_responses", ascending=False)
)

display(duplicate_instructions.head(20))
```

```python
print(
    "Duplicated instructions with multiple responses:",
    (duplicate_instructions["n_responses"] > 1).sum()
)

print(
    "Duplicated instructions with identical responses:",
    (duplicate_instructions["n_responses"] == 1).sum()
)
```

```python
intent_category_map = (
    df.groupby("intent")["category"]
      .nunique()
      .sort_values(ascending=False)
)

display(intent_category_map)
```
